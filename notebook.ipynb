### Algoritmo de Divisão

#### [1.1] Teste os exemplos

Exemplos:  $5 \mid 15; 6 \not\mid 15; 0 \mid 0; 1 \mid -1$

```python
def DIV(b, a):
    if b == 0:
        return a == 0
    return a % b == 0

print("5 é divisor de 15" if DIV(5, 15) else "5 não é divisor de 15")
print("6 é divisor de 15" if DIV(6, 15) else "6 não é divisor de 15")
print("0 é divisor de 0" if DIV(0, 0) else "0 não é divisor de 0")
print("1 é divisor de -1" if DIV(1, -1) else "1 não é divisor de -1")
```

Para testar $0 \mid 0$, recorremos à definição formal, uma vez que a operação $0$ % $0$ retorna erro por computar o resto de uma divisão por $0$. Dessa forma, a partir da definição, sabemos que, para que $0$ seja divisor de $0$, precisamos da existência de algum $q$ tal que $q \cdot 0 = 0$. Dessa forma, qualquer valor de $q$ satisfaz a condição, ressultando em $0 = 0$ e garantindo que $0$ seja divisor de $0$

#### [1.2] Teste estas propriedades.

**Propriedades**
$$
\begin{array}{ll}
i)\quad & c \mid b \wedge b \mid a ~\Rightarrow ~ c \mid a \\
ii)\quad & b \mid a \wedge b \mid a' ~\Rightarrow ~ b \mid a + a' \\
iii)\quad & b \mid 0 ~~\forall b \in \mathbb{Z} \\
iv)\quad & 1 \mid a ~~\forall a \in \mathbb{Z} \\
v) &0 \mid a \iff a=0 \\
\end{array}
$$

```python
amostra = list(range(-30, 31))

print("Testes das propriedades")
print("-" * 50)

prop_i_valida = True
for c in amostra:
    for b in amostra:
        for a in amostra:
            if DIV(c, b) and DIV(b, a):
                if not DIV(c, a):
                    prop_i_valida = False

print(f"Propriedade i)  (Transitividade): {prop_i_valida}")

prop_ii_valida = True
for b in amostra:
    for a in amostra:
        for a_linha in amostra:
            if DIV(b, a) and DIV(b, a_linha):
                if not DIV(b, a + a_linha):
                    prop_ii_valida = False

print(f"Propriedade ii) (Linearidade):    {prop_ii_valida}")


prop_iii_valida = all(DIV(b, 0) for b in amostra)
print(f"Propriedade iii) (Divisores de 0): {prop_iii_valida}")

prop_iv_valida = all(DIV(1, a) for a in amostra)
print(f"Propriedade iv)  (Divisão por 1):  {prop_iv_valida}")

prop_v_valida = True
for a in amostra:
    if DIV(0, a) != (a == 0):
        prop_v_valida = False

print(f"Propriedade v)   (Múltiplos de 0): {prop_v_valida}")
```

#### [2.4] Use o Teorema dos Números Primos para estimar quantos números primos existem com tamanhos de 1024 bits, 1536 bits e 2048 bits (os tamanhos usados na criptografia)

Um número de $k$ bits se encontra no intervalo definido entre $2^{k-1}$ e $2^k -1$. Para descobrir quantos primos existem nesse intervalo, ou seja, quantos primos possuem $k$ bits, pegamos o número de primos até o limite inferior $2^{k-1}$ e subtraímos do número de primos até o limite superior $2^k -1$.

Como na criptografia trabalhamos com números muito altos, podemos considerar que o comportamento assintótico é atingido. Assim, 

$$\frac{\pi(N)}{N}\approx \frac{1}{\ln(N)}$$

Multiplicando ambos os lados por $N$:

$$\pi(N) \approx \frac{N}{\ln(N)}$$

Dessa forma, podemos determinar a aproximação dos limites superiores e inferiores.

- Primos até o limite inferior ($2^{k-1}$):

$$\pi(2^{k-1}) \approx \frac{2^{k-1}}{\ln(2^{k-1})} = \frac{2^{k-1}}{(k-1)\ln(2)} $$

- Primos até o limite superior ($2^k -1$):

$$\pi(2^k -1 ) \approx \frac{2^k}{\ln(2^k)} = \frac{2^{k}}{(k)\ln(2)} $$

Concluimos que o número de números primos que existem com $k$ bits é:

$$\text{Total de primos com k bits} \approx \frac{2^{k}}{k \ln(2)} - \frac{2^{k-1}}{(k-1) \ln(2)} = \frac{2^{k-1}}{\ln(2)} \cdot \left( \frac{k-2}{k^2 - k} \right)$$

Podemos converter isso em uma função em python para aproximar os valores:


```python
import math
from decimal import Decimal, getcontext

getcontext().prec = 10

def aproximar_primos(k):
    k_dec = Decimal(k)
    ln2 = Decimal(math.log(2))
    
    potencia = (Decimal(2) ** (k - 1)) / ln2
    fracao = (k_dec - 2) / (k_dec**2 - k_dec)
    
    total = potencia * fracao
    return total

print(f"Total para 1024 bits: {aproximar_primos(1024):.3e}")
print(f"Total para 1536 bits: {aproximar_primos(1536):.3e}")
print(f"Total para 2048 bits: {aproximar_primos(2048):.3e}")
```

Portanto, descobrimos que o total de números primos são:
- 1024 bits: $1.265 \times 10^{305}$
- 1536 bits: $1.131 \times 10^{459}$
- 2048 bits: $1.138 \times 10^{613}$

#### [2.6] Construa um número primo com seu nome

```python
import random

def fermat_primality_test(n, k=10):
		"""
		Teste de primalidade de Fermat para verificar se n é provavelmente primo.

		Argumentos:
				n (int): O número a ser testado.
				k (int): O número de testes aleatórios a serem realizados (padrão=10).

		Retorna:
				bool: Verdadeiro se n passar em todos os testes (provavelmente primo), Falso se composto.
"""
		if n <= 1:
				return False
		elif n <= 3:
				return True # 2 e 3 são primos
		elif n % 2 == 0:
				return False # Números pares >2 são compostos

		for _ in range(k):
				a = random.randint(2, n - 2) # Escolha um valor aleatório a onde 1 < a < n-1
				if pow(a, n - 1, n) != 1: # Calcula a^(n-1) mod n eficientemente
						return False # Definitivamente composto
		return True # Provavelmente primo

def converter_frase(frase):
    """Converte letras de A-Z para 01-26 e espaços para 27."""
    numero_str = ""
    for letra in frase.upper():
        if 'A' <= letra <= 'Z':
            valor = ord(letra) - 64
            # O :02d garante que o número sempre terá 2 dígitos (ex: 1 vira 01)
            numero_str += f"{valor:02d}" 
        elif letra == ' ':
            numero_str += "27"
    return numero_str
```

#### [3.1] Escreva uma função ver_cpf() que verifica se um CPF é válido (retornar True) ou não (retornar False)

```python
def verifica_primeiro_digito(cpf):
    dig_cpf = [int(numero) for numero in cpf] ## transformo a string cpf em um vetor de número para os cálculos
    produto = 1
    somatorio = 0
    for i in range(len(dig_cpf) - 2): ## percorre de 0 a 8
        somatorio = somatorio + (produto * dig_cpf[i])
        produto += 1

    resultado = somatorio % 11 #modulo 11

    if resultado == 10:
        resultado = 0
        return resultado
    else:
        return resultado
        


def verifica_ultimo_digito(cpf):
    dig_cpf = [int(numero) for numero in cpf] ## transformo a string cpf em um vetor de número para os cálculos
    produto = 0
    somatorio = 0
    for i in range(len(dig_cpf) - 1): ## percorre de 0 a 9
        somatorio = somatorio + (produto * dig_cpf[i])
        produto += 1
    
    resultado = somatorio % 11 #modulo 11
    
    if resultado == 10:
        resultado = 0
        return resultado
    else:
        return resultado
    
```

```python
def ver_cpf(cpf):
    dig_cpf = [int(numero) for numero in cpf] ## transformamos em vetor para facilitar

    if dig_cpf[9] == verifica_primeiro_digito(cpf) and dig_cpf[10] == verifica_ultimo_digito(cpf):
        print("Válido")
        return
    else:
        print("Inválido")
        return
```

#### [3.2] Verifique este exemplo usando Python. Use as seguintes funções para converter entre letras e números

```python
# Versão  usando string e index()
def numero_para_letra_v2(numero):
    """Converte um número (0-25) para letra (A-Z) usando string e index."""
    letras = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
    if not 0 <= numero <= 25:
        raise ValueError("O número deve estar entre 0 e 25.")
    return letras[numero]

def letra_para_numero_v2(letra):
    """Converte uma letra (A-Z) para número (0-25) usando string e index()."""
    letras = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
    if not letra.isalpha() or not letra.isupper() or len(letra) != 1:
        raise ValueError("A entrada deve ser uma única letra maiúscula (A-Z).")
    return letras.index(letra)
```

```python
def descriptografador(chave, cifrado):
    vet_chave = list(chave) ## cada posição do vetor passa a ter uma letra
    vet_cifrado = list(cifrado) ## cada posição do vetor passa a ter uma letra

    número_da_descrip = [0] * len(cifrado) ## armazenar os números das letras
    letras_da_descrip = list(cifrado)

    for i in range(len(vet_cifrado)):
        número_da_descrip[i] = (letra_para_numero_v2(vet_cifrado[i]) - letra_para_numero_v2(vet_chave[i])) % 26

    for i in range(len(número_da_descrip)):
       letras_da_descrip[i] = numero_para_letra_v2(número_da_descrip[i])
    
    letras_da_descrip = "".join(letras_da_descrip)
    print(letras_da_descrip)
```

```python
def calcula_MDC(a, b): 
    while b != 0:
        q = a // b
        r = a % b
        (a, b) = (b, a%b)
    return a

def checa_coprimos(n1,n2,n3):
    if calcula_MDC(n1,n2) == 1 and calcula_MDC(n1,n3) == 1 and calcula_MDC(n2,n3) == 1:
        return True
    return False

def calcula_n(a,b,c):
    return a * b * c

def calcula_sol_parcial(num, mod, n1, n2):
    for k in range(1,mod):
        teste = n1 * n2 * k
        if (teste % mod == num):
            return teste
        
def encontra_solucao(a1, a2, a3, n1, n2, n3):
    if not checa_coprimos(n1,n2,n3):
        raise ValueError("Náo säo coprimos!")
    n = calcula_n(n1,n2,n3)
    sol1 = calcula_sol_parcial(a1, n1, n2, n3)
    sol2 = calcula_sol_parcial(a2, n2, n1, n3)
    sol3 = calcula_sol_parcial(a3, n3, n1, n2)
    return (sol1 + sol2+ sol3) % n

solucao = encontra_solucao(5, 4, 10, 7, 11, 13)

print("O resultado encontrado foi:", solucao)
print("Checando resultado...")
print(solucao, "mod(7) = 5?", solucao % 7 == 5)
print(solucao, "mod(11) = 4?", solucao % 11 == 4)
print(solucao, "mod(13) = 10?", solucao % 13 == 10)


```

```python
def euclides_estendido(a, b):
    if b == 0:
        return a, 1, 0
    else:
        mdc, x1, y1 = euclides_estendido(b, a % b) ## chamando recursivamente
        x = y1
        y = x1 - (a // b) * y1
        return mdc, x, y
```

#### 5.2 Identifique quais outros elementos módulo 11 são geradores e quais não são.

```python
def ordem(a, p):
    """Calcula a ordem de a em Z_p* (menor k>0 tal que a^k ≡ 1 mod p)."""
    for k in range(1, p):
        if pow(a, k, p) == 1:
            return k

p = 11

print()
print(f"Ordens dos elementos de Z_{p}*:")
geradores = []
for a in range(1, p):
    ord_a = ordem(a, p)
    eh_gerador = (ord_a == p - 1)
    simbolo = "GERADOR" if eh_gerador else f"ordem {ord_a}"
    print(f"a = {a:<3}  ord(a) = {ord_a:<4}  {simbolo}")
    if eh_gerador:
        geradores.append(a)

print(f"\nGeradores de Z_{p}*: {geradores}")
```

##### Protocolo DHKE


1. Alice e Bob concordam com um número primo grande $p$ e um gerador $g$. 
Eles podem até encontrar bons exemplos na internet, e seus valores não precisam ser ocultados.

2. Alice gera um número aleatório $a \in \mathbb{Z}_{p-1}^+$, calcula $A = g^a \pmod{p}$. 
Ela envia $A$ para Bob, mantendo $a$ em segredo.

3.  Agora Bob faz o mesmo: ele gera um número aleatório $b \in \mathbb{Z}_{p-1}^+$, 
calcula $B = g^b \pmod{p}$. Ele envia $B$ para Alice, mantendo $b$ em segredo.

4. Ao receber a mensagem $B$, Alice calcula sua chave $k_A = B^a \pmod{p}$.

5. Ao receber a mensagem $A$, Bob calcula sua chave $k_B = A^b \pmod{p}$.


Isso completa o protocolo básico. Observe que Alice e Bob agora compartilham a mesma 
chave porque:

\begin{align*}
k_A &\equiv B^a \equiv (g^b)^a \equiv g^{ba} \equiv g^{ab} \equiv (g^a)^b \equiv A^b \equiv k_B \pmod{p}
\end{align*}

Mesmo que o adversário veja $A = g^a$ e $B = g^b$, ele não pode calcular a chave $k$. 
É verdade que ele pode calcular $AB = (g^a)(g^b) = g^{a+b}$, mas $g^{a+b} \neq g^{ab} = k$.


\begin{array}{ccc}
\textbf{Alice} & & \textbf{Bob} \\
\hline
& \xleftarrow{\displaystyle p,\ g} & \\
a \in \mathbb{Z}_{p-1}^+ & & b \in \mathbb{Z}_{p-1}^+ \\
A = g^a & & B = g^b \\
& \xrightarrow{\displaystyle A} & \\
& \xleftarrow{\displaystyle B} & \\
k_A = B^a & & k_B = A^b \\
\end{array}


#### 5.5 Implemente uma versão simplificada do protocolo DHKE usando $p = 2^{64} - 2^{32} + 1$ e $g = 7$. Simule os passos de Alice e Bob acima e verifique se $k_A \equiv k_B$.

```python
p = 2**64 - 2**32 + 1   # primo de Solinas (muito usado em criptografia)
g = 7
import random

# Alice escolhe a e calcula A = g^a mod p
a = random.randint(1, p - 2)     
A = pow(g, a, p)                 
print(f"Alice:")
print(f"Chave privada a = {a}")

# Bob escolhe b e calcula B = g^b mod p 
b = random.randint(1, p - 2)     
B = pow(g, b, p)                  
print(f"Bob:")
print(f"Chave privada b = {b}")

# Cada um calcula a chave compartilhada
kA = pow(B, a, p)   # Alice calcula k = B^a mod p
kB = pow(A, b, p)   # Bob calcula   k = A^b mod p

print(f"\nChaves calculadas:")
print(f"Alice => k_A = {kA}")
print(f"Bob   => k_B = {kB}")
print(f"k_A == k_B? {kA == kB}  ✓" if kA == kB else f"k_A == k_B? ERRO!")
```
